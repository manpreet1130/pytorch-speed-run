### Reshaping, Stacking, Squeezing and Unsqueezing
- reshaping: change input tensor shape to desired shape
- view: new tensor has the original tensor's address, any changes to new happen to old
- hstack: horizontal stack
- vstack: vertical stack
- squeeze: remove all 1 dimensions from a tensor
- unsqueeze: add a 1 dimension to a target tensor
- permute: return input where dimensions are swapped in a certain way

```python
tensor = torch.arange(1., 10.)
tensor, tensor.shape
```

```python
tensor_reshaped = tensor.reshape(3, 3)
tensor_reshaped, tensor_reshaped.shape
```

```python
tensor_view = tensor.view(9, 1)
tensor_view, tensor_view.shape
```

```python
# stacking
tensor_stack = torch.stack([tensor, tensor, tensor], dim = 1)
tensor.shape, tensor_stack, tensor_stack.shape
```

```python
tensor_vstack = torch.vstack([tensor, tensor, tensor])
tensor_vstack
```

```python
tensor_hstack = torch.hstack([tensor, tensor, tensor])
tensor_hstack
```

```python
# squeeze
tensor = torch.reshape(tensor, (1, 1, 9))
tensor, tensor.shape
```

```python
tensor_squeezed = torch.squeeze(tensor)
tensor_squeezed, tensor_squeezed.shape
```

```python
tensor_unsqueezed = torch.unsqueeze(tensor_squeezed, dim = 0)
tensor_unsqueezed = torch.unsqueeze(tensor_unsqueezed, dim = 2)
tensor_unsqueezed, tensor_unsqueezed.shape
```

```python
# torch.permute - it a view, so same memory as original tensor
tensor = torch.rand((224, 224, 3))
permuted_tensor = tensor.permute(2, 0, 1)
tensor.shape, permuted_tensor.shape
```

## PyTorch and NumPy




```python
import numpy as np

arr = np.arange(1, 8)
# numpy default is float64 but pytorch default is float32
tensor = torch.from_numpy(arr).type(torch.float32)
arr, tensor, tensor.dtype
```

```python
# tensor to numpy
tensor = torch.ones(7)
# numpy default is float64, pytorch default is float32
numpy_tensor = tensor.numpy().astype(np.float64)
tensor, numpy_tensor, numpy_tensor.dtype
```

### Reproducibility

```python
# create two random tensors
a = torch.rand(3, 4)
b = torch.rand(3, 4)
print(a)
print(b)
print(a == b)
```

```python
# set a random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
a = torch.rand(3, 4)
torch.manual_seed(RANDOM_SEED)
b = torch.rand(3, 4)
print(a)
print(b)
print(a == b)
```
